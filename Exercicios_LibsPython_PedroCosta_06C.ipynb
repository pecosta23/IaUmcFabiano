## Exercicio 1

Crie um array unidimensional de 50 elementos contendo numeros inteiros aleatorios entre 10 e 100.


```python
array = np.array(np.random.randint(10, 101, size=(50))) #variavel array contem o randint de 10 ate 100, com 50, unidimensional
print(array)
print(len(array)) #prova real para ver se tinha acertado
print(min(array)) #valor minimo atual
print(max(array)) #valor maximo
```

## Exercicio 2

Gere uma matriz 4x4 com numeros inteiros de 1 a 16 e extraia a ultima linha.


```python
array = np.array(np.random.randint(1, 17, size=(4, 4))) # mesma coisa do exe acima, mas com 1 até 16 random, matriz 4x4
print(f' Array: \n{array}\n') # ver o total
ultima_linha = array[-1, :] # '-1, :' pega a ultima linha, no pdf de exemplo tinhamos o de primeira linha: '[0, :]'
print(f'ultima linha: {ultima_linha}')
```

## Exercicio 3

Crie um array de zeros com comprimento 15 e substitua os valores da 5a a 10a posicao por 1.


```python
array = np.array(np.zeros(15)) #gera o array uni com 15 zeros
print(array)
array[4:10] = 1 #posicao de 5 ate 10 nao indice de 5 ate 10
print(array)
```

## Exercicio 4

Gere um array de 100 numeros inteiros aleatorios e encontre a mediana e a moda dos valores.


```python
array = np.array(np.random.randint(0, 101, size=100)) #aleatorio, com 100 uni
print(array)
mediana = np.median(array) #calcula a mediana
print(mediana)
moda = np.bincount(array).argmax() #moda
print(moda)
```

## Exercicio 5

Crie um array de 20 elementos contendo numeros entre 1 e 100 e substitua todos os valores multiplos de 3 por -3.


```python
array = np.array(np.random.randint(1, 101, size=20)) # aleatorio entre 1 ate 100, com 20 uni
print(array)
for n in array: #listando cada indice
    if n % 3 == 0: #mod de 3, se for multiplo
        array[array == n] = -3 #substitui
print(array) 
```

## Exercicio 6

Gere uma matriz 5x5 de numeros aleatorios entre 0 e 1 e encontre a soma de cada coluna.


```python
array = np.array(np.random.randint(0, 2, size=(5, 5))) #matriz 5x5 aleteatorio entre 0 e 1
print(array)
for col in range(array.shape[1]): #col ve o tamanho do array e circula ele
    soma = np.sum(array[:, col]) # ':' pega do indice 0 até 'col' que é o tamanho do array
    print(f'coluna {col}: {soma}')
```

## Exercicio 7

Crie dois arrays de numeros aleatorios e encontre a correlacao entre eles.


```python
array1 = np.array(np.random.randint(0, 10, size=(3))) #testei com 5, 2, e 3. 5 ficou dificil de entender 2 ficou muito binário, 3 ficou bom
array2 = np.array(np.random.randint(0, 10, size=(3)))
print(f'\n{array1}\n')
print(f'\n{array2}\n')

correlacao = np.corrcoef(array1, array2) #tinha usado antes a funcao correlate, porque n sabia o nome da certa
print(f'correlacao: \n{correlacao}\n') # correlacao muda quando gero os arrays, tenho indicativos de alta e baixa -1 ate 1

'''Legal observar que quando uma sobe a outra desce (correlacao)'''

#correlacao = np.correlate(array1, array2, mode='full')
#print(f'\ncorrelacao: {correlacao}\n')
```

## Exercicio 8

Gere uma matriz 3x3 com numeros inteiros aleatorios e eleve todos os seus elementos ao quadrado.


```python
array = np.array(np.random.randint(1, 11, size=(3, 3)))
print(array)
for linha in range(0, 3):
    for coluna in range(0, 3):
        quadrado = array[linha][coluna] ** 2

print(quadrado)
```
