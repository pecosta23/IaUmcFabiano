## Exercicio 9

Crie um array com 10 numeros aleatorios e normalize seus valores entre 0 e 1.


```python
array = np.array(np.random.rand(10))
print(array)
max = np.max(array)

min = np.min(array)

print(f'\nMax: {max}\nMin: {min}\n')

lista = []

for n in array:

    normalize = (n - min) / (max - min)

    lista.append(normalize)

array_final = np.array(lista)
    

print(f'Array normalizado: {array_final}')
```

## Exercicio 10

Crie uma matriz identidade 6x6 e substitua a diagonal principal por valores crescentes de 1 a 6.


```python
array = np.eye(6, 6, order='C')

for linha in range(0, 6):
    for coluna in range(0, 6):
        if linha == coluna:
            array[linha, coluna] = linha + 1

print(array)
```

## Exercicio 19

Crie dois arrays de numeros inteiros aleatorios de tamanho 20 e calcule o angulo entre eles (produto escalar normalizado).


```python
array1 = np.array(np.random.randint(1, 21, size=20))
array2 = np.array(np.random.randint(1, 21, size=20))

produto_escalar = 0
print(f'{array1}\n{array2}\n')

for n, j in zip(array1, array2):
    conta = n * j
    produto_escalar += conta

print(f'Produto escalar: {produto_escalar}')

somaArray1 = 0
somaArray2 = 0
for n in array1:
    somaArray1 += n ** 2

for n in array2:
    somaArray2 += n ** 2

normaArray1 = somaArray1 ** 0.5
normaArray2 = somaArray2 ** 0.5

print(f'Normaa do array 1: {normaArray1}\nNorma do array 2: {normaArray2}')

produto_escalar_normalizado = produto_escalar / (normaArray1 * normaArray2)

print(f'Produto escalar normalizado: {produto_escalar_normalizado}')

def fatorial(i):
    conta = 1
    for n in range(i, 0, -1):
        conta = conta * n 
    return conta

def coseno_serie_taylor(x):
    resultado = 0
    for n in range(0, 100, 2):
        termo = (x ** n) / fatorial(n)
        if (n // 2) % 2 == 0:
            resultado = resultado + termo
        else:
            resultado = resultado - termo

    return resultado

menor_valor = float("inf")
melhor_angulo = 0
for angulo in range(181):

    radianos = angulo * (3.14/180)

    cos = coseno_serie_taylor(radianos)

    diferenca = cos - produto_escalar_normalizado


    if diferenca < 0:
        diferenca = diferenca * -1

    if diferenca < menor_valor:
        menor_valor = diferenca
        melhor_angulo = angulo
            
print(f'Melhor angulo !!!!!!! HAHAAHAHAHAHA: {melhor_angulo}')
print(f'Menor diferenca: {menor_valor}')
```

## Exercicio 20

Crie uma matriz de numeros inteiros e encontre o valor minimo de cada linha.


```python
array = np.array(np.random.randint(1, 11, size=(5,5)))
print(array)
for linha in array:
    menor = linha[0]

    for n in linha:
        if n < menor:
            menor = n

    print(menor)
```
