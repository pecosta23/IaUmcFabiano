## 4. Conversão de Temperatura

**Enunciado:** Converta uma temperatura fornecida em Celsius para Fahrenheit usando a fórmula F = C * 9/5 + 32.

```python
temp = float(input('Digite a temperatura em Celcius: ')) # entrada celcius
fahren = (temp * 9/5) + 32 # processamento, priorizando multiplicacao
print(f"A temperatura equivalalente em Farenheit é: {fahren}") # saída f string
```

## 5. Tabuada

**Enunciado:** Crie um programa que imprima a tabuada de um número fornecido pelo usuário.

```python
num = int(input('Digite um número para saber sua tabuada: ')) #entrada
print(f"Tabuada do número {num}") #UX lol
for n in range (1, 11): # loop simples, n queria mostrar o 0 que é o primeioro indice, ent coloquei 1, 11
  conta = num * n # entrada do user e o loop até 10
  print(f"{num} * {n} = {conta}") # f string com os dados
```

## 8. Verificação de Palíndromo

**Enunciado:** Crie um programa que verifique se uma palavra é um palíndromo (ex: "radar").

```python
palavra = input("Digite a palabra: ") #entrada da palavra
ehpalin = palavra[::-1] # usei slicing para ver o padrao da palavra, se for um palindromo a input e o slicing vao ter resultados iguais
if palavra == ehpalin: # comparacao
  print('Essa palavra é um palíndromo') # ele, radar, rever, subi no onibus
else: # resto
  print('Náo é um palíndromo')
```
