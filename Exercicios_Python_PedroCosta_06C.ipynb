## 21. Busca Binária

**Enunciado:** Implemente a busca binária para encontrar um elemento em uma lista ordenada.

```python
# busca binária, pega do meio identifica se o valor é >= ou <= repartindo até encontrar
lista = [ -1, 1, 2, 4, 6, 7, 9, 14, 15, 16, 19, 20, 24, 27, 32, 37, 41, 43, 67] # lista com vários numeros ordenados em posicoes diferentes

user = int(input('Qual número devemos buscar? ')) # entrada do usuário

indiceInicio = 0 # tenho que colocar o inicio da lista no indice 0
indiceFim = len(lista) - 1 # mesma coisa mas para o final, assim o algoritmo sabe em qual espaço procurar

while indiceInicio <= indiceFim: #enquanto sempre true
  indiceMeio = (indiceInicio + indiceFim) // 2 #acha o meio por // para n ter float

  if lista[indiceMeio] == user: #quando achar
    print(f"Número na posição {indiceMeio} encontrado")
    break

  elif lista[indiceMeio] < user: #n achou busca na frente
    indiceInicio = indiceMeio + 1

  else:
    indiceFim = indiceMeio - 1 #n achou busca atrás

if user not in lista: #fallback se n estiver na lista
  print("Esse número não está na lista")
```

## 22. Criptografia de Cifra de César

**Enunciado:** Crie uma função para criptografar uma mensagem usando a cifra de César.

```python
deslocamento = 0 #instancio com um valor inicial

def criptografar(user, deslocamento): #crio a funcao com input da frase e o deslocamento
  resultado = "" # crio uma variavel de str vazia para preencher

  for n in user: # passa por todos os caracteres da frase do usuario
    if n.isalpha(): # se for letra entrar, qq outra coisa n
      num = ord(n) #veriavel num recebe o index do for (letra) transforma em numero
      criptografia = num + deslocamento # var cript pega o numero e soma o deslocamento
      letra = chr(criptografia) # pego o resultado e transforma em letra
      resultado += letra #letra compoem a frase final que é a var resultado
    else:
      resultado += n #diferente de letra cai no else e é ignorado

  return resultado
```

## 37. Funções Recursivas

**Enunciado:** Implemente uma função recursiva para calcular o fatorial de um número.

```python
# funcao que chama ela mesma
def fatorial(n): # entrada do usuario
    if n == 0 or n == 1: # caso base, se for 0 ou 1 o fatorial é 1
        return 1
    else:
        return n * fatorial(n - 1) #recursao, onde diminui o valor do n até chegar no caso base


num = int(input("Digite um num: "))

resultado = fatorial(num)

print(f"fatorial de {num} eh {resultado}")
```

**Enunciado:** Crie um programa para gerenciar uma lista de compras, permitindo adicionar, remover e listar itens.

```python
compras = []

while True:
    print("1 - Adicionar")
    print("2 - Remover")
    print("3 - Listar")
    print("4 - Sair")

    opcao = input("Escolha um numero de 1 ate 4: ")

    if opcao == "1":
        item = input("qual eh oitem que deseja adicionar: ")
        compras.append(item) # adc a lista
        print("Item adicionado!")

    elif opcao == "2":
        item = input("qual eh  o item que deseja remover: ")

        if item in compras:
            compras.remove(item) #remove da lista
            print("item removido")
        else: #fallback
            print("esse item não está na lista.")

    elif opcao == "3":
        print("Lista: ")

        for item in compras:
            print(f"- {item}") # mostra a lista

    elif opcao == "4":
        print("saindo") #sai do loop
        break

    else:
        print("digte um numero de 1 ate 4")
```

## 46. Calcular Média e Desvio Padrão

**Enunciado:** Calcule a média e o desvio padrão de uma lista de números.

```python
lista = [9, 3, 37, 21, 85]
# Calculando a média
soma = 0
for numero in lista:
    soma += numero #soma dos numeros da lista

media = soma / len(lista)

#dsvio padrao
quadrados = 0

for numero in lista:
    diferenca = numero - media # dif entre num e a media da lista
    quadrados += diferenca ** 2 # soma dos quadrados das diferencas

variancia = quadrados / len(lista)
desvio_padrao = variancia ** 0.5 # raiz quadrada da variancia, poderia ter usado sqrt(), mas n quis importar

print(f"media: {media}")
print(f"desvio padrao: {desvio_padrao}")
```

## 48. Ordenação por Inserção

**Enunciado:** Implemente o algoritmo de ordenação por inserção (Insertion Sort).

```python
lista = [5, 2, 4, 6, 1, 3] #lista aletoria

for i in range(1, len(lista)): # range do tamanho da lista
    chave = lista[i] # var recebe o indice
    j = i - 1 # j recebe o indice anterior

    while j >= 0 and lista[j] > chave: # enquanto condicao verdadeira, desloca o indice
        lista[j + 1] = lista[j]
        j -= 1 # retira um indice

    lista[j + 1] = chave #valor da chave recebe o indice

print(lista)
```

## 49. Soma dos Dígitos

**Enunciado:** Calcule a soma dos dígitos de um número inteiro fornecido.

```python
num = int(input("Digite um número inteiro: "))
soma = 0

for digito in str(abs(num)): #converte em string e adc sobre cada indice, abs() lida com num negativos
    soma += int(digito) #adiciona o valor

print("soma dos digitos:", soma)
```

## 50. Cálculo de Frequência de Letras

**Enunciado:** Calcule a frequência de cada letra em uma string fornecida pelo usuário.

```python
# msm ideia do exe 34 e de 47

frase = input("Digite uma frase: ")
frequencia = {} #dicionario das letras

for letra in frase.lower(): #padroniza
    if letra.isalpha(): #confere se eh letra
        if letra in frequencia:
            frequencia[letra] += 1 #adc primeiro indice
        else:
            frequencia[letra] = 1 #adc

print(frequencia)
```
